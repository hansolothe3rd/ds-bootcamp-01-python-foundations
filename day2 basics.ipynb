### 1. Basic example:

```python
temp = 85
if temp > 90:
    print("It's hot!")
elif temp > 70:
    print("It's nice out.")
else:
    print("It's chilly.")
```

### 2. Challenge: Write a program that checks if a number is positive, negative, or zero.

```python
# Ask the user to enter a number
number = float(input("Enter a number: "))

# Check if the number is positive, negative, or zero
if number > 0:
    print("The number is positive.")
elif number < 0:
    print("The number is negative.")
else:
    print("The number is zero.")
```

## Step 4: Functions with Logic

### 1. Function with multiple conditions:

```python
def classify_number(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"

print(classify_number(10))
print(classify_number(-5))
print(classify_number(0))
```

### Write a program that:

1. Asks for a user’s age (use input()).

2. If age < 13 → print “Child.”
If 13–19 → print “Teenager.”
If 20+ → print “Adult.”

3. Wrap this in a function age_group(age).

```python
# 1. Asks for a user’s age (use input()).

# Ask the user for their age
age = input("Please enter your age: ")

# Print it back to them
print(f"You are {age} years old.")
```

```python
# 2. If age < 13 → print “Child.” If 13–19 → print “Teenager.” If 20+ → print “Adult.”

# Ask the user for their age
age = int(input("Please enter your age: "))

# Check the age category
if age < 13:
    print("Child.")
elif 13 <= age <= 19:
    print("Teenager.")
else:
    print("Adult.")
```

```python
# 3. Wrap this in a function age_group(age).

def age_group(age):
    if age < 13:
        return "Child."
    elif 13 <= age <= 19:
        return "Teenager."
    else:
        return "Adult."


# Example usage:
age = int(input("Please enter your age: "))
print(age_group(age))
```

```python
def age_group(age):
    if age < 13:
        return "Child."
    elif 13 <= age <= 19:
        return "Teenager."
    else:
        return "Adult."


# Unit tests
def test_age_group():
    assert age_group(5) == "Child."
    assert age_group(12) == "Child."
    assert age_group(13) == "Teenager."
    assert age_group(17) == "Teenager."
    assert age_group(19) == "Teenager."
    assert age_group(20) == "Adult."
    assert age_group(30) == "Adult."
    print("All tests passed!")


# Run tests
test_age_group()
```

## Step 6: Mini Project – FizzBuzz

### The classic programming challenge:

* Print numbers from 1 to 50.

* For multiples of 3, print "Fizz".

* For multiples of 5, print "Buzz".

* For multiples of both 3 and 5, print "FizzBuzz".

```python
for number in range(1, 51):  # 1 to 50 inclusive
    if number % 3 == 0 and number % 5 == 0:
        print("FizzBuzz")
    elif number % 3 == 0:
        print("Fizz")
    elif number % 5 == 0:
        print("Buzz")
    else:
        print(number)
```
