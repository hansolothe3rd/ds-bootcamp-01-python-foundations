## Step 1: Lists (review + deeper dive)
Lists are ordered, mutable collections.

```python
# Creating a list
numbers = [10, 20, 30, 40, 50]

# Indexing & slicing
print(numbers[0])      # first element
print(numbers[-1])     # last element
print(numbers[1:4])    # slice [20, 30, 40]

# Modifying
numbers.append(60)     # add at end
numbers.insert(2, 25)  # insert at index 2
numbers.remove(40)     # remove value
```

## Step 3: Sets
Sets are unordered, unique collections.

```python
fruits = {"apple", "banana", "cherry", "apple"}
print(fruits)   # duplicates removed

# Set operations
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))        # {1, 2, 3, 4, 5}
print(a.intersection(b)) # {3}
```

1. Word Counter

```python
# Step 1: Write a function that takes a sentence and returns a dictionary with each word and its count. 
#         Example: "data science is fun and data is powerful"
#         → {'data': 2, 'science': 1, 'is': 2, 'fun': 1, 'and': 1, 'powerful': 1}

def word_count(sentence):
    # Split the sentence into words
    words = sentence.split()
    
    # Create an empty dictionary to store counts
    counts = {}
    
    # Loop through each word and count
    for word in words:
        if word in counts:
            counts[word] += 1
        else:
            counts[word] = 1
            
    return counts

# Example usage
sentence = "data science is fun and data is powerful"
print(word_count(sentence))
```

💡 Tip: You can also do this in one line using Python’s collections.Counter:

```python
from collections import Counter

sentence = "data science is fun and data is powerful"
counts = Counter(sentence.split())
print(counts)
```

2. Unique Items

```python
#    Write a function that takes a list and returns a set of unique items.

def unique_items(lst):
    # Convert the list to a set to remove duplicates
    return set(lst)

# Example usage
my_list = [1, 2, 3, 2, 4, 1, 5]
print(unique_items(my_list))
```

Here’s a version that removes duplicates but keeps the original order:

```python
def unique_items_ordered(lst):
    unique = []
    seen = set()
    for item in lst:
        if item not in seen:
            unique.append(item)  # add to list in order
            seen.add(item)       # remember we’ve seen it
    return unique

# Example usage
my_list = [1, 2, 3, 2, 4, 1, 5]
print(unique_items_ordered(my_list))
```

## Step 6: Mini Project – Phone Book
Build a simple phone book app using a dictionary:

```python
phone_book = {}

# Add contacts
phone_book["Sierra"] = "555-1234"
phone_book["Luna"] = "555-5678"

# Lookup
name = "Sierra"
if name in phone_book:
    print(f"{name}'s number is {phone_book[name]}")
else:
    print(f"{name} not found")
```

👉 Expand it:

* Add more contacts.

* Write a function to search, add, and delete contacts.

```python
# Step 3: Define functions

def search_contact(name):
    """Search for a contact by name."""
    return phone_book.get(name, "Contact not found.")

def add_contact(name, number):
    """Add a new contact."""
    phone_book[name] = number
    print(f"Added {name}: {number}")

def delete_contact(name):
    """Delete a contact by name."""
    if name in phone_book:
        del phone_book[name]
        print(f"Deleted {name}.")
    else:
        print("Contact not found.")

# Step 4: Example usage
print("\n📒 Current phone book:", phone_book)

print("\n🔍 Search:")
print(search_contact("Sierra"))
print(search_contact("Dan"))

print("\n➕ Add:")
add_contact("Dan", "555-9999")
print(phone_book)

print("\n❌ Delete:")
delete_contact("Luna")
print(phone_book)
```

Let’s turn it into a simple interactive phone book program where you can keep adding, searching, and deleting contacts until you choose to quit.

```python
# 📒 Interactive Phone Book

# Start with some contacts
phone_book = {
    "Sierra": "555-1234",
    "Luna": "555-5678"
}

# Define functions
def search_contact(name):
    """Search for a contact by name."""
    return phone_book.get(name, "Contact not found.")

def add_contact(name, number):
    """Add a new contact."""
    phone_book[name] = number
    print(f"Added {name}: {number}")

def delete_contact(name):
    """Delete a contact by name."""
    if name in phone_book:
        del phone_book[name]
        print(f"Deleted {name}.")
    else:
        print("Contact not found.")

# Menu loop
while True:
    print("\n📞 Phone Book Menu")
    print("1. Search Contact")
    print("2. Add Contact")
    print("3. Delete Contact")
    print("4. Show All Contacts")
    print("5. Quit")

    choice = input("Choose an option (1-5): ")

    if choice == "1":
        name = input("Enter name to search: ")
        print(search_contact(name))

    elif choice == "2":
        name = input("Enter name: ")
        number = input("Enter phone number: ")
        add_contact(name, number)

    elif choice == "3":
        name = input("Enter name to delete: ")
        delete_contact(name)

    elif choice == "4":
        print("📒 All Contacts:", phone_book)

    elif choice == "5":
        print("Goodbye! 👋")
        break

    else:
        print("Invalid choice. Please try again.")
```
