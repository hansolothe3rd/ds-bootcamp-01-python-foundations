⚓ Titanic Survival Analysis (Beginner Version)

1. What percentage of passengers survived?

### 1. What percentage of passengers survived?
Here’s a step-by-step example:

```python
import pandas as pd

# Step 1: Load the Titanic dataset
# If you don't have it, you can download from: https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)

# Step 2: Preview the data
print(titanic.head())

# Step 3: Calculate the percentage of passengers who survived
total_passengers = len(titanic)
survived_passengers = titanic['Survived'].sum()  # 'Survived' column: 1 = survived, 0 = did not
survival_percentage = (survived_passengers / total_passengers) * 100

print(f"Total passengers: {total_passengers}")
print(f"Number survived: {survived_passengers}")
print(f"Percentage survived: {survival_percentage:.2f}%")
```

### 2. Compare survival rates between men and women.

```python
# Group by Sex and calculate survival rate
survival_by_sex = titanic.groupby("Sex")["Survived"].mean() * 100

print("Survival rates by gender:")
print(survival_by_sex)
```

### 3. Find the average age of survivors vs non-survivors.

```python
# Group by survival status and calculate mean age
avg_age = titanic.groupby("Survived")["Age"].mean()

print("Average age of survivors vs non-survivors:")
print(avg_age)
```

### 4. How many children (<12) survived?

```python
# Filter children under 12
children = titanic[titanic["Age"] < 12]

# Count how many survived
children_survived = children["Survived"].sum()
children_total = len(children)

print(f"Children under 12: {children_total}")
print(f"Children who survived: {children_survived}")
print(f"Survival rate: {children_survived / children_total * 100:.2f}%")
```
