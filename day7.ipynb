### 🧑‍💻 Step 1: Load Your Dataset

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)

# Inspect
titanic.head()
titanic.info()
```

### 📌 Step 2: Column Operations

```python
# Create a new column: family size
titanic['FamilySize'] = titanic['SibSp'] + titanic['Parch']

# Create a binary column: is_child (<12 years)
titanic['IsChild'] = titanic['Age'] < 12

# Create a new column using a condition
titanic['HighFare'] = titanic['Fare'] > 50

titanic[['Name','FamilySize','IsChild','HighFare']].head()
```

### 📌 Step 3: Aggregation & Value Counts

```python
# Count survivors by class
print(titanic.groupby('Pclass')['Survived'].sum())

# Survival rate by sex
print(titanic.groupby('Sex')['Survived'].mean())

# Count unique values
print(titanic['Embarked'].value_counts())
```

### 📌 Step 4: Filtering & Sorting

```python
# Children who survived
children_survived = titanic[(titanic['IsChild']) & (titanic['Survived']==1)]
print(children_survived[['Name','Age','Survived']])

# Top 10 highest fares
top_fares = titanic.sort_values('Fare', ascending=False).head(10)
print(top_fares[['Name','Fare','Survived']])
```

### 📌 Step 5: Mini Project – Titanic Insights

1. How does family size relate to survival?

2. How many children (<12) survived, and what was their class distribution?

3. How many passengers had high fares (>50) and survived?

4. What is the most common Embarked port among survivors?

### 1. How does family size relate to survival?


```python
# Create FamilySize feature
titanic["FamilySize"] = titanic["SibSp"] + titanic["Parch"] + 1
#(where +1 counts the passenger themselves).

# Group by family size and calculate survival rate
family_survival = titanic.groupby("FamilySize")["Survived"].mean() * 100

print("Survival rate by family size:")
print(family_survival)
```

### 2. How many children (<12) survived, and what was their class distribution?

```python
# Filter children under 12
children = titanic[titanic["Age"] < 12]

# Count survivors
children_survived = children[children["Survived"] == 1]

print(f"Total children under 12: {len(children)}")
print(f"Children who survived: {len(children_survived)}")

# Class distribution among survivors
class_distribution = children_survived["Pclass"].value_counts()

print("\nClass distribution of child survivors:")
print(class_distribution)

# Group by class and calculate survival rate
class_survival = children.groupby("Pclass")["Survived"].mean() * 100

print("Survival rate of children (<12) within each class (%):")
print(class_survival)
```

### 3. How many passengers had high fares (>50) and survived?

```python

# Count survivors among HighFare passengers
highfare_survived = titanic[(titanic["HighFare"] == True) & (titanic["Survived"] == 1)]

print(f"Total HighFare passengers: {titanic['HighFare'].sum()}")
print(f"HighFare passengers who survived: {len(highfare_survived)}")

# Survival rate for HighFare passengers
survival_rate = len(highfare_survived) / titanic["HighFare"].sum() * 100
print(f"Survival rate for HighFare passengers: {survival_rate:.1f}%")
```

### 4. What is the most common Embarked port among survivors?

```python
# Filter survivors
survivors = titanic[titanic["Survived"] == 1]

# Count embarkation ports among survivors
embarked_counts = survivors["Embarked"].value_counts()

print("Most common Embarked port among survivors:")
print(embarked_counts.head(1))  # top 1
```
