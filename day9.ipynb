## 🔧 Step-by-Step Instructions
### 1. Imports

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load Titanic
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)

# Inspect
titanic.head()
titanic.info()
```

### 2. Boxplots & Violin Plots

These help visualize distributions across categories.

```python
# Boxplot: Fare by Class
sns.boxplot(x="Pclass", y="Fare", data=titanic)
plt.title("Fare Distribution by Passenger Class")
plt.show()

# Violin plot: Age by Survival
sns.violinplot(x="Survived", y="Age", data=titanic, split=True)
plt.title("Age Distribution by Survival")
plt.show()
```

### 3. Swarm Plot (individual points + distribution)

```python
plt.figure(figsize=(10,6))
sns.swarmplot(x="Pclass", y="Age", hue="Survived", data=titanic, dodge=True, palette="Set1")

plt.title("Age Distribution in Each Class by Survival", fontsize=16, weight="bold")
plt.xlabel("Passenger Class", fontsize=12)
plt.ylabel("Age", fontsize=12)
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

### 4. Pairplot (scatter matrix)

Visualizes multiple relationships at once.

```python
sns.pairplot(
    titanic[["Age", "Fare", "Pclass", "Survived"]],
    hue="Survived"
)
plt.show()
```

### 5. Heatmap (correlations)

```python
plt.figure(figsize=(8,6))
sns.heatmap(titanic.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap – Titanic Data")
plt.show()
```

### 6. Subplots with Matplotlib

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

sns.histplot(titanic["Age"].dropna(), bins=20, ax=axes[0], color="blue")
axes[0].set_title("Age Distribution")

sns.histplot(titanic["Fare"].dropna(), bins=20, ax=axes[1], color="green")
axes[1].set_title("Fare Distribution")

plt.tight_layout()
plt.show()
```
